```python
import json
import pandas as pd
```

```python
with open("countries-50m.json") as json_file:
    json_countries = json.load(json_file)
```

```python
#json_countries.get("objects").get("countries").get("geometries")

country_names_json = []
country_id_json = []
for country_entry in json_countries.get("objects").get("countries").get("geometries"):
    country_names_json.append(country_entry.get("properties").get("name"))
    country_id_json.append(country_entry.get("id"))

len(country_names_json)
```

```python
pd.DataFrame({"country_json": country_names_json, "id_json": country_id_json})\
    .sort_values(by="country_json")\
    .to_csv("countries_id_json.csv", index=False)
```
